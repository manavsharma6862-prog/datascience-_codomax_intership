```python
import pandas as pd

# Load the raw dataset
df = pd.read_csv('students.csv')

print("Original Dataset:")
print(df)
print("Shape:", df.shape)

# Step 1: Remove duplicate rows
df_clean = df.drop_duplicates()

# Step 2: Handle missing values in 'Marks' column (fill with column mean)
mean_marks = round(df_clean['Marks'].mean(), 2)
df_clean['Marks'] = df_clean['Marks'].fillna(mean_marks)

print("\nCleaned Dataset:")
print(df_clean)

# Step 3: Export cleaned dataset to a new CSV file
df_clean.to_csv('students_cleaned.csv', index=False)

print("\nCleaned dataset saved as 'students_cleaned.csv'")
```
