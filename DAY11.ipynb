```python
#Import Pandas
import pandas as pd
```

```python
#Load Dataset
df = pd.read_csv("students.csv")

print(df)
```

```python
#Total Number of Students
total_students = df["Name"].count()

print("Total Students:", total_students)
```

```python
#Average Marks
average_marks = df["Marks"].mean()

print("Average Marks:", average_marks)
```

```python
#. Student with Highest Marks
highest = df[df["Marks"] == df["Marks"].max()]

print(highest)
```

```python
 #Student with Lowest Marks
lowest = df[df["Marks"] == df["Marks"].min()]

print(lowest)
```

```python
#Course-wise Average Marks
course_average = df.groupby("Course")["Marks"].mean()

print(course_average)
```

```python
#. Generate Business Insights
print("===== BUSINESS INSIGHTS =====")

print("1. Total Students:", df["Name"].count())

print("2. Average Marks:", round(df["Marks"].mean(), 2))

print("3. Highest Marks:", df["Marks"].max())

print("4. Lowest Marks:", df["Marks"].min())

print("5. Most Popular Course:", df["Course"].mode()[0])

print("=============================")
```
