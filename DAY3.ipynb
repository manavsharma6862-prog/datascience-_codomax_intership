```python
1. #Create a 1D Array
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("1D Array:")
print(arr)
```

```python
2.#Create a 2D Array
import numpy as np

arr = np.array([[1, 2, 3],
                [4, 5, 6]])

print("2D Array:")
print(arr)
```

```python
3. #Array Indexing
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("First Element:", arr[0])
print("Last Element:", arr[-1])
```

```python
4.# Array Slicing
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr[1:4])
```

```python
5. #Addition of Two Arrays
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Addition:")
print(a + b)
```

```python
6.# Subtraction, Multiplication and Division
import numpy as np

a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
7. #Reshape an Array
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

new_arr = arr.reshape(2, 3)

print(new_arr)
```
