```python
#Calculate Total Marks
import pandas as pd

df = pd.read_csv("students.csv")

total_marks = df["Marks"].sum()

print("Total Marks:", total_marks)
```

```python
#Calculate Average Marks
import pandas as pd

df = pd.read_csv("students.csv")

average_marks = df["Marks"].mean()

print("Average Marks:", average_marks)
```
