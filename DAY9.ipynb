```python
#Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Load Dataset
df = pd.read_csv("students.csv")

print(df.head())
```

```python
#VDisplay Basic Information
print("Dataset Information")
print(df.info())

print("\nStatistical Summary")
print(df.describe())
```

```python
#Business Insights
print("Total Students:", df["Name"].count())
print("Average Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())
```

```python
#. Bar Chart
plt.figure(figsize=(6,4))

plt.bar(df["Name"], df["Marks"])
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

```python
#Line Chart
plt.figure(figsize=(6,4))

plt.plot(df["Name"], df["Marks"], marker="o")

plt.title("Marks Trend")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

```python
#Pie Chart
plt.figure(figsize=(6,6))

plt.pie(df["Marks"],
        labels=df["Name"],
        autopct="%1.1f%%",
        startangle=90)

plt.title("Marks Distribution")

plt.show()
```

```python
#Dashboard Summary
print("========== MINI DASHBOARD ==========")
print("Total Students :", df["Name"].count())
print("Average Marks  :", df["Marks"].mean())
print("Highest Marks  :", df["Marks"].max())
print("Lowest Marks   :", df["Marks"].min())
print("====================================")
```
